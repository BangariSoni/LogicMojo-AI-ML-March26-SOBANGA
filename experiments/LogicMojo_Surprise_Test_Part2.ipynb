# Question 2

```python
data={
    'Area':[1000,1500,1800,2200,2500],
    'Bedrooms':[2,3,3,4,4,],
    'Price':[50,65,72,85,95]
}
df=pd.DataFrame(data)
print(df)
```

```python
X=df[['Area','Bedrooms']]
print(X)
```

```python
Y=df['Price']
print(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
print("Coefficients:",model.coef_)

print("Intercept:",model.intercept_)

print("MSE:",mean_squared_error(Y_test,y_pred))

print("R2:",r2_score(Y_test,y_pred))
```

```python
new_house=[[2000,3]]

print("Predicted Price:",model.predict(new_house))
```

# Question 3

```python
X=np.array([1,2,3,4,5])
X=X.reshape(-1,1)
print(X)
y = np.array([2, 8, 18, 32, 50])
```

```python
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)
```

```python
print(X_poly)
```

```python
model = LinearRegression()

model.fit(X_poly, y)
```

```python
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
y_pred=model.predict(X_poly)
print(y_pred)
```

```python
print("Coefficients:",model.coef_)

print("Intercept:",model.intercept_)

print("MSE:",mean_squared_error(y,y_pred))

print("R2:",r2_score(y,y_pred))
```

```python
X_new = poly.transform([[6]])

prediction = model.predict(X_new)

print(prediction)
```

# Question 6

```python
scores = np.array([45, 50, 55, 60, 65, 70, 75])
mean = np.mean(scores)
std = np.std(scores)      
z_score = (75 - mean) / std
print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-score of 75:", z_score)
if abs(z_score) > 2:
    print("75 is an Outlier")
else:
    print("75 is NOT an Outlier beacuse it is less than 2")
```

```python
mean = 70
std = 8
prob_between = norm.cdf(78, mean, std) - norm.cdf(62, mean, std)
prob_above = 1 - norm.cdf(86, mean, std)
print("Between 62 and 78:", prob_between)
print("Above 86:", prob_above)
```

# Question 10

```python
before = np.array([70, 72, 68, 75, 74])
after = np.array([78, 80, 76, 82, 79])
alpha = 0.05
t_statistic, p_value = ttest_rel(after, before)
print("T-statistic:",t_statistic)
print("P-value:",p_value)
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```
