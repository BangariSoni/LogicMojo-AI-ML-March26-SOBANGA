Q1. Derivative of f(x) = x³·sin(x)
Answer: C — 3x²sin(x) + x³cos(x)
This uses the Product Rule: if f(x) = u·v, then f'(x) = u'v + uv'

u = x³ → u' = 3x²
v = sin(x) → v' = cos(x)
f'(x) = 3x²·sin(x) + x³·cos(x)


Q2. Primary objective of Gradient Descent
Answer: B — Minimize the loss function by updating parameters in the negative gradient direction
Gradient tells you which direction the loss increases. So you move in the opposite (negative) direction to reduce loss. This is how models learn — by repeatedly nudging weights to reduce error.

Q3. Fastest growing function as x → ∞
Answer: D — eˣ
Growth order (slowest to fastest):

log(x) << x << x² << eˣ

Exponential always wins over any polynomial or logarithm at large x.

Q4. Derivative of f(x) = ln(x)
Answer: A — 1/x
This is a standard result. The natural log grows slowly, and its derivative 1/x keeps shrinking — which makes sense geometrically. Used heavily in cross-entropy loss.

Q5. What does the gradient vector represent geometrically?
Answer: B — Direction of steepest ascent where the function increases fastest
The gradient points uphill (steepest increase). That's why Gradient Descent goes in the negative gradient direction — to go downhill toward minimum loss.

Q6. Rule used for differentiating composite functions f(g(x))
Answer: C — Chain Rule
Chain Rule: d/dx[f(g(x))] = f'(g(x)) · g'(x)
This is the backbone of backpropagation — each layer's gradient is multiplied through using chain rule.

Q7. Condition that indicates a function is convex
Answer: B — The second derivative remains positive

f''(x) > 0 means the curve bends upward → convex → has a single global minimum
This is why convex loss functions (like MSE) are preferred — no local minima traps


Q8. Derivative of f(x) = eˣ
Answer: C — eˣ
The exponential function is its own derivative. This unique property makes it ideal in activation functions and probability models.

Q9. Matrix containing all second-order partial derivatives
Answer: C — Hessian Matrix
MatrixContainsJacobianFirst-order partial derivativesHessianSecond-order partial derivativesIdentityDiagonal 1s
Hessian helps determine if a critical point is a min, max, or saddle point.

Q10. Dot product of orthogonal vectors
Answer: C — 0
Dot product = |A||B|cos(θ). When θ = 90°, cos(90°) = 0, so dot product = 0.
This is fundamental in PCA, where principal components are orthogonal to each other.

Q11. Matrix decomposition most closely associated with PCA
Answer: C — Singular Value Decomposition (SVD)
PCA finds directions of maximum variance. SVD decomposes a matrix into U, Σ, Vᵀ — the singular vectors directly give you the principal components. Most PCA implementations use SVD under the hood.

Q12. Why is the determinant important?
Answer: B — It determines whether a matrix is invertible or singular

det(A) ≠ 0 → matrix is invertible → unique solution exists
det(A) = 0 → matrix is singular → cannot be inverted → no unique solution

Critical when solving linear systems in ML (e.g., Normal Equation in Linear Regression).

Q13. Mathematical expression for mutually exclusive events
Answer: B — P(A ∩ B) = 0
Mutually exclusive = they cannot both happen. So their intersection (both occurring) has probability 0. Example: a coin flip can't be both heads and tails.

Q14. Gradient vector of f(x, y) = x² + y²
Answer: B — (2x, 2y)
Gradient = vector of partial derivatives:

∂f/∂x = 2x
∂f/∂y = 2y
∇f = (2x, 2y)

This bowl-shaped function is a classic convex example used in teaching gradient descent.

Q15. Gradients becoming extremely small during backpropagation
Answer: B — Vanishing Gradient Problem
In deep networks, when gradients are multiplied through many layers (chain rule), they can shrink toward zero — early layers stop learning. This is why ReLU activation and batch normalization were introduced as solutions.

Q16. Definite integral ∫₀¹ x² dx
Answer: B — 1/3
∫x² dx = x³/3
Evaluate from 0 to 1: (1³/3) - (0³/3) = 1/3

Q17. Most accurate interpretation of an eigenvector
Answer: B — A vector whose direction remains unchanged after transformation
When you apply a matrix transformation to an eigenvector, it only scales (by the eigenvalue λ) — it doesn't rotate. In PCA, eigenvectors of the covariance matrix are the principal components.

Q18. Probability as degree of belief based on evidence
Answer: C — Bayesian Interpretation
InterpretationViewFrequentistProbability = long-run frequency of eventsBayesianProbability = degree of belief, updated with evidenceClassicalEqual likelihood outcomes
Bayesian ML updates beliefs using Bayes' Theorem as new data arrives.

Q19. Derivative of f(x) = cos(x)
Answer: B — −sin(x)
Standard trig derivatives:

d/dx[sin(x)] = cos(x)
d/dx[cos(x)] = −sin(x)

The negative sign is important — many students miss it.

Q20. Local minima vs Global minima
Answer: C — Local minima is smallest in a nearby region; global minima is smallest overall

Local minimum: lowest point in its neighborhood — a "valley"
Global minimum: the absolute lowest point across the entire function

In non-convex neural network loss surfaces, Gradient Descent can get stuck in local minima — which is why techniques like momentum, Adam optimizer, and random restarts are used.
