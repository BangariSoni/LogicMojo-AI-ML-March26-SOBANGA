```python
#Squares of a Sorted Array o/p [0,1,9,16,100]
arr = [-4,-1,0,3,10]

result = [0] * len(arr)

left = 0
right = len(arr) - 1
pos = len(arr) - 1

while left <= right:

    if abs(arr[left]) > abs(arr[right]):
        result[pos] = arr[left] ** 2
        left += 1
    else:
        result[pos] = arr[right] ** 2
        right -= 1

    pos -= 1

print(result)
```

```python
def sorted_squares(arr):
    n = len(arr)
    result = [0] * n          # pre-fill result array
    left  = 0
    right = n - 1
    pos   = n - 1             # fill result from RIGHT to LEFT

    while left <= right:
        left_sq  = arr[left]  ** 2
        right_sq = arr[right] ** 2

        if left_sq > right_sq:
            result[pos] = left_sq
            left += 1
        else:
            result[pos] = right_sq
            right -= 1

        pos -= 1              # move position leftward

    return result

arr = [-4, -1, 0, 3, 10]
print(sorted_squares(arr))    # [0, 1, 9, 16, 100] ✅
```

```python
#Merge Sorted Array o/p:[1,2,2,3,5,6]
# nums1 = [1,2,3,0,0,0]
# m = 3
# nums2 = [2,5,6]
# n = 3
def merge(nums1, m, nums2, n):
    p1 = m - 1          # pointer for nums1 valid elements
    p2 = n - 1          # pointer for nums2
    p  = m + n - 1      # pointer for end of nums1

    while p1 >= 0 and p2 >= 0:
        if nums1[p1] > nums2[p2]:
            nums1[p] = nums1[p1]   # place larger at end
            p1 -= 1
        else:
            nums1[p] = nums2[p2]
            p2 -= 1
        p -= 1

    # if nums2 still has elements remaining
    while p2 >= 0:
        nums1[p] = nums2[p2]
        p2 -= 1
        p  -= 1

    return nums1

nums1 = [1, 2, 3, 0, 0, 0]
nums2 = [2, 5, 6]
print(merge(nums1, 3, nums2, 3))   # [1, 2, 2, 3, 5, 6] ✅
```

```python
#Sort colors (Dutch National Flag Problem) o/p[0,0,1,1,2,2]
#i/p:[2,0,2,1,1,0]
def sortColors(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:

        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1

        elif arr[mid] == 1:
            mid += 1

        else:  # arr[mid] == 2
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1

    return arr
sortColors([2,0,2,1,1,0])
```

```python
#Container With Most Water
def max_water(height):
    left  = 0
    right = len(height) - 1
    max_water = 0

    while left < right:
        # water = min of two heights × distance between them
        w= right-left
        h=min(height[left], height[right])
        water = w* h
        max_water = max(max_water, water)

        # move the shorter side inward
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return max_water

height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(max_water(height))  
```

```python
#3 Sum ///total sum of 2 numbers should be 0 that is a+b+c=0
arr=nums = [-1,0,1,2,-1,-4]
def threeSum(nums: list[int]):
        result = []
        nums.sort()
        n = len(nums)

        for i in range(n - 2):
            # No three positive numbers can sum to zero
            if nums[i] > 0:
                break

            # Skip duplicate values for the first element
            if i > 0 and nums[i] == nums[i - 1]:
                continue

            left, right = i + 1, n - 1

            while left < right:
                total = nums[i] + nums[left] + nums[right]

                if total < 0:
                    left += 1
                elif total > 0:
                    right -= 1
                else:
                    result.append([nums[i], nums[left], nums[right]])
                    # Skip duplicates for both pointers
                    while left < right and nums[left] == nums[left + 1]:
                        left += 1
                    while left < right and nums[right] == nums[right - 1]:
                        right -= 1
                    left += 1
                    right -= 1
                    
        return result
threeSum([-1,0,1,2,-1,-4])
    
```
